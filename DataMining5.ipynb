Problem-06
Using a simulated two-dimensional dataset:
i.	Perform Agglomerative Hierarchical Clustering using Ward linkage. 
ii.	Construct and display a dendrogram using scipy.cluster.hierarchy.
iii.	 From the dendrogram, determine the appropriate number of clusters. 
iv.	Assign observations to the identified clusters.
v.	Calculate the ARI and interpret the result

```python


from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import adjusted_rand_score
from scipy.cluster.hierarchy import dendrogram,linkage

```

```python
#Generate 2D dataset
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
X,y_true=make_blobs(n_samples=100,
                    centers=3,
                    n_features=2,
                    random_state=35)
```

```python
#i)Agglomerative Hierarchical Clustering Using Ward Linkage
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt
Z=linkage(X,method="ward")
print(Z)

```

```python
#ii)construct dendrogram
plt.figure(figsize=(12,5))
dendrogram(Z)
plt.title("Dendrogram")
plt.xlabel("Samples")
plt.ylabel("Euclidean Distance")
plt.show()
```

```python
#iv)
from sklearn.cluster import AgglomerativeClustering
model = AgglomerativeClustering(n_clusters=3, linkage='ward')
y_pred = model.fit_predict(X)
print(y_pred)
```

```python
# Plot identified clusters
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='viridis')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Agglomerative Hierarchical Clustering')
plt.show()
```

```python
# 5. Calculate ARI
from sklearn.metrics import adjusted_rand_score
ari = adjusted_rand_score(y_true, y_pred)
print("Adjusted Rand Index (ARI):", round(ari, 4))
```

Interpretation: Agglomerative Hierarchical Clustering
Agglomerative Hierarchical Clustering was applied to the dataset using Ward's linkage method to identify three clusters. The Adjusted Rand Index (ARI) was 0.833, indicating a high level of agreement between the predicted clusters and the true cluster labels. This result suggests that the clustering algorithm successfully identified the original group structure, although some observations may have been assigned to different clusters. Overall, the model demonstrated good clustering performance.
