Problem-04


Generate a two-dimensional clustering dataset using make_blobs to perform K-Means clustering.
i.	Plot the generated observations.
ii.	Apply K-Means with 3 Clusters. 
iii.	Plot the resulting clusters and cluster centroids. 
iv.	Calculate the Adjusted Rand Index (ARI) using the true cluster labels.
v.	Interpret the obtained ARI value.


```python
# Import libraries
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
```

```python
# Generate 2D dataset
X, y_true = make_blobs(
    n_samples=300,
    centers=3,
    n_features=2,
    random_state=35
)
```

```python
# i) Plot generated observations
plt.scatter(X[:, 0], X[:, 1])
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Generated Dataset')
plt.show()
```

```python
# ii Apply K-Means with 3 clusters
kmeans = KMeans(
    n_clusters=3,
    random_state=35,
    n_init=10
)

y_pred = kmeans.fit_predict(X)
```

```python
# iii) Plot resulting clusters and centroids
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='viridis')

centroids = kmeans.cluster_centers_

plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker='X',
    s=200,
    color='red',
    label='Centroids'
)

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('K-Means Clustering (K=3)')
plt.legend()
plt.show()
```

```python
# iv) Calculate Adjusted Rand Index
ari = adjusted_rand_score(y_true, y_pred)

print("Adjusted Rand Index (ARI):", round(ari, 4))
```

Interpretation of K-Means Clustering
The K-Means algorithm identified three distinct clusters (K = 3) in the generated dataset. The red crosses represent the centroids of the three clusters. The Adjusted Rand Index (ARI) was 0.9605, indicating excellent agreement between the predicted clusters and the true cluster labels. Overall, the results show that K-Means successfully identified the original clusters with very high accuracy.
